# Import library

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Import all log CSVs

```python
monte_carlo_logs = pd.read_csv("logs/montecarlo_visual_training.csv")
sarsa_logs = pd.read_csv("logs/sarsa_visual_training.csv")
qlearning_logs = pd.read_csv("logs/qlearning_training.csv")
dqn_logs = pd.read_csv("logs/dqn_visual_training.csv")
ppo_logs = pd.read_csv("logs/ppo_visual_training.csv")
```

# Episode vs Collisions

```python
fig, axs = plt.subplots(2, 2, figsize=(40, 20))
axs[0, 0].plot(sarsa_logs['episode'], sarsa_logs['collisions'], label='SARSA', color='orange')
axs[0, 0].set_title('SARSA Training Collisions')
axs[0, 1].plot(qlearning_logs['episode'], qlearning_logs['collisions'], label='Q-Learning', color='green')
axs[0, 1].set_title('Q-Learning Training Collisions')
axs[1, 0].plot(dqn_logs['episode'], dqn_logs['collisions'], label='DQN', color='red')
axs[1, 0].set_title('DQN Training Collisions')
axs[1, 1].plot(monte_carlo_logs['episode'], monte_carlo_logs['collisions'], label='Monte Carlo', color='purple')
axs[1, 1].set_title('Monte Carlo Training Collisions')
```

# Episode vs Rewards

```python
fig, axs = plt.subplots(2, 2, figsize=(40, 20))
axs[0, 0].plot(sarsa_logs['episode'], sarsa_logs['reward'], label='SARSA', color='orange')
axs[0, 0].set_title('SARSA Training Reward')
axs[0, 1].plot(qlearning_logs['episode'], qlearning_logs['reward'], label='Q-Learning', color='green')
axs[0, 1].set_title('Q-Learning Training Reward')
axs[1, 0].plot(dqn_logs['episode'], dqn_logs['reward'], label='DQN', color='red')
axs[1, 0].set_title('DQN Training Reward')
axs[1, 1].plot(monte_carlo_logs['episode'], monte_carlo_logs['reward'], label='Monte Carlo', color='purple')
axs[1, 1].set_title('Monte Carlo Training Reward')
```

# Episode vs Steps

```python
fig, axs = plt.subplots(2, 2, figsize=(40, 20))
axs[0, 0].plot(sarsa_logs['episode'], sarsa_logs['steps'], label='SARSA', color='orange')
axs[0, 0].set_title('SARSA Training Steps')
axs[0, 1].plot(qlearning_logs['episode'], qlearning_logs['steps'], label='Q-Learning', color='green')
axs[0, 1].set_title('Q-Learning Training Steps')
axs[1, 0].plot(dqn_logs['episode'], dqn_logs['steps'], label='DQN', color='red')
axs[1, 0].set_title('DQN Training Steps')
axs[1, 1].plot(monte_carlo_logs['episode'], monte_carlo_logs['steps'], label='Monte Carlo', color='purple')
axs[1, 1].set_title('Monte Carlo Training Steps')
```
